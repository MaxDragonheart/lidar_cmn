# Download dei dati

## Dati iniziali

```python
import logging
from pathlib import Path

from lidar_cmn.config import log_format

logging.basicConfig(level=logging.INFO, format=log_format)

lidar_type = ["dtm", "dsm"]

target_area_name = "Procida"

output_folder = Path.home().joinpath("lidar")
```

## Selezione e Download

```python
from lidar_cmn.vector import read_boundaries, read_lidar_index, target_area

boundaries = read_boundaries()

for lidar in lidar_type:
    base = read_lidar_index(lidar_type=lidar)
    ax = base.plot(alpha=0.75, color="blue", figsize=(10, 10))
    ax.set_title(lidar)

    data = target_area(
        boundaries=boundaries,
        link_source=base,
        target=target_area_name
    )
    data.plot(ax=ax, alpha=0.75, color="yellow")
```

### DTM

```python
from lidar_cmn.download import download_data

dtm = target_area(
    boundaries=boundaries,
    link_source=read_lidar_index(lidar_type=lidar_type[0]),
    target=target_area_name
)

download_data(
    selection=dtm,
    output_folder=output_folder,
    output_folder_name=lidar_type[0]
)
```

### DSM

```python
dsm = target_area(
    boundaries=boundaries,
    link_source=read_lidar_index(lidar_type=lidar_type[1]),
    target=target_area_name
)

download_data(
    selection=dsm,
    output_folder=output_folder,
    output_folder_name=lidar_type[1]
)
```

## Download dell'intero dataset
Scarica tutti i 5451 tasselli del DTM e del DSM: impostare `download_all = True`. Se il download si interrompe, basta rieseguire la cella: i file già scaricati vengono saltati.

```python
download_all = False

if download_all:
    for lidar in lidar_type:
        download_data(
            selection=read_lidar_index(lidar_type=lidar),
            output_folder=output_folder,
            output_folder_name=lidar
        )
```
